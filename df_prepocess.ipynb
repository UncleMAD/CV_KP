```python
def create_small_dataset(df, frac=0.2, random_state=42):
    return (
        df.groupby('Classes', group_keys=False)
          .apply(lambda x: x.sample(frac=frac, random_state=random_state))
          .reset_index(drop=True)
    )
```

```python
train_small = create_small_dataset(pd.read_csv("data/fgvc-aircraft/train.csv"))
val_small = create_small_dataset(pd.read_csv("data/fgvc-aircraft/val.csv"))
test_small = create_small_dataset(pd.read_csv("data/fgvc-aircraft/test.csv"))

train_small.to_csv('data/fgvc-aircraft/train_small.csv', index=False)
val_small.to_csv('data/fgvc-aircraft/val_small.csv', index=False)
test_small.to_csv('data/fgvc-aircraft/test_small.csv', index=False)
```
